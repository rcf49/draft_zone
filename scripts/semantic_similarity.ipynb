# Natual Language Processing (NLP)

## Word2Vec

* Vectorize sentence using general English library dataset
* More general results, since using a general library
* Using spaCy

[![Example vector](../images/example_vector.png)](https://spacy.io/usage/spacy-101#vectors-similarity)

## Term Frequency-Inverse Document Frequency (TF-IDF)

* Vectorize sentence using the frequency of terms (in the sentence and in the entire dataset)
* More use case specific results, since it considers the frequency of terms in the dataset
* Using Scikit Learn

[![TF-IDF graph](../images/tfidf.png)](https://medium.com/data-science/tfidf-for-piece-of-text-in-python-43feccaa74f8)

## Cosine-Similarity

* Use cosine-similarity to calculate the similarity between sentences
* Works best on short, single sentences
* Ranges from 0-1
    * 1 is exact match between the vectors

[![Example vector](../images/cosine_similarity.png)](https://www.oreilly.com/library/view/statistics-for-machine/9781788295758/eb9cd609-e44a-40a2-9c3a-f16fc4f5289a.xhtml)

```python
import pandas as pd
import spacy
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from tqdm import tqdm
from pprint import pprint
nlp = spacy.load("en_core_web_lg")
```

```python
# Function to prepare string for spacy similarity
def spacy_prep(text):
    # Tokenize string
    doc = nlp(text)

    # Remove stop words and punctuation
    prepped = [token.text for token in doc if (not token.is_stop) and (not token.is_punct)]

    # Then re-tokenize
    final = nlp(str(" ".join(prepped)))
    return final
```

```python
# Dataframe of sample text
df = pd.DataFrame({"Text": ["First example sentence", "Oh look, another example", "I like cheese!"]})

# Prepare text for spacy
df['spacy_text'] = df["Text"].apply(spacy_prep)

df.head()
```

```python
# Spacy similarities
spacy_array = list()
for row_doc in df["spacy_text"]:
    row = list()
    for other_doc in df["spacy_text"]:
        word2vec = row_doc.similarity(other_doc)
        row.append(word2vec)
    spacy_array.append(row)
pprint(spacy_array)
```

```python
# Sckit Learn vectors and similarities
vectorizer = TfidfVectorizer(stop_words="english")
tfidf = vectorizer.fit_transform(df["Text"])
scikit_similarity = cosine_similarity(tfidf.toarray())

pprint(scikit_similarity)
```
